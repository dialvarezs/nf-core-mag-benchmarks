```python
from pathlib import Path

import polars as pl
from plotnine import ggplot
```

```python
data_bins = pl.concat(
    [
        pl.read_csv(
            p / "bin_summary.tsv",
            separator="\t",
            infer_schema_length=None,
            null_values=[""],
        ).with_columns(dataset=pl.lit(p.name))
        for p in sorted(Path("../data").iterdir())
        if p.is_dir()
    ],
    # los datasets zymo traen columnas checkm2/gtdbtk que maghini no tiene,
    # y algunas columnas comunes difieren en Int64/Float64 entre corridas
    how="diagonal_relaxed",
)
```
